```python
import torch
from diffusers import StableAudioPipeline

pipe = StableAudioPipeline.from_pretrained(
    "stabilityai/stable-audio-open-1.0", torch_dtype=torch.float16
).to("cuda")
print("model ready")
```

```python
import os, torch, soundfile as sf

os.makedirs("sounds", exist_ok=True)

sounds = [
    ("a soft rising whoosh of water, something drifting upward toward a distant glow", 7),
    ("a single soft underwater heartbeat thump, deep and round", 3),
    ("a deep low ocean hum, calm and slowly shifting, very spacious", 15),
    ("rapid light clicking and scraping of thin tentacles in dark water", 5),
    ("a slow flowing water current with gentle bubbles rising", 12),
    ("a sharp electric crackle and a short stinging buzz, like a warning", 4),
    ("shimmering glassy chimes, many tiny bells ringing softly underwater", 9),
    ("a gentle tidal swell, calm waves, a soft low tone beneath", 14),
    ("a near-silent deep drone with a faint sub-bass breath", 13),
    ("a steady soft pulsing thump, slow and rhythmic, calming", 8),
    ("glittering water sparkle, high bubbling shimmer in the dark", 10),
    ("a swarm of tiny wet pops and rustles, many small creatures drifting", 11),
    ("a warm glowing tone like a humming glass harmonica underwater", 6),
    ("soft wet slurping and gulping sounds, like something drinking the water", 4.5),
    ("a quiet low pulse with a faint distant rumble", 3.5),
    ("a slow descending water whoosh settling into calm stillness", 10.5),
]

for i, (prompt, secs) in enumerate(sounds):
    generator = torch.Generator("cuda").manual_seed(i)
    audio = pipe(
        prompt,
        negative_prompt="Low quality.",
        num_inference_steps=100,
        audio_end_in_s=secs,
        generator=generator,
    ).audios[0]
    sf.write(f"sounds/sound_{i:02d}.wav", audio.T.float().cpu().numpy(), pipe.vae.sampling_rate)
    print("saved", i, secs, "s")
```
